<i>Understanding outliers</i>

- Separating non-binary columns

```python
binary_cols = [col for col in dataset.select_dtypes(["int64", "float64"]).columns 
               if dataset[col].nunique() == 2]
continuous_cols = [col for col in dataset.select_dtypes(["int64", "float64"]).columns 
                    if col not in binary_cols and col != "label"]
```

- Finding outliers

```python
for col in continuous_cols:
    q1=dataset[col].quantile(0.25)
    q3=dataset[col].quantile(0.75)
    iqr=q3-q1
    lower=q1-1.5*iqr
    upper=q3+1.5*iqr
    outliers=((dataset[col]<lower)|(dataset[col]>upper))
    print(f"For {col}:",outliers.sum()/len(dataset)*100)
```

<i>Handling outliers</i>

```python
for col in continuous_cols:
    q1=dataset[col].quantile(0.25)
    q3=dataset[col].quantile(0.75)
    iqr=q3-q1
    lower=q1-1.5*iqr
    upper=q3+1.5*iqr
    dataset[col]=np.where(dataset[col]<lower,lower,np.where(dataset[col]>upper,upper,dataset[col]))
    outliers=((dataset[col]<lower)|(dataset[col]>upper))
    print(f"For {col}:",outliers.sum()/len(dataset)*100)
```

<i>Splitting the dataset into Matrix of Features (X) and Independent Variable (y)</i>

```python
X=dataset.iloc[:,:-1].values
y=dataset.iloc[:,-1].values
```

<i>Splitting it into training and test set</i>

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

<i>Feature scaling</i>

```python
from sklearn.preprocessing import StandardScaler

feature_cols = dataset.drop("label", axis="columns").columns
continuous_indexes = [feature_cols.get_loc(col) for col in continuous_cols]
continuous_indexes = feature_cols.get_indexer(continuous_cols)

scaler=StandardScaler()
X_train[:, continuous_indexes] = scaler.fit_transform(X_train[:, continuous_indexes])
X_test[:, continuous_indexes] = scaler.transform(X_test[:, continuous_indexes])
```

2.<u>Model builiding</u>

<b>a. Logistic Regression</b>

<i>Model building</i>

```python
from sklearn.linear_model import LogisticRegression

classifier1=LogisticRegression()
classifier1.fit(X_train,y_train)
```

<i>Predicting the values for training and test set</i>

```python
y_pred_train1=classifier1.predict(X_train)
y_pred_test1=classifier1.predict(X_test)

print("For the training set:")
print("[Actual_observations Predicted_observations]")
print(np.concatenate((y_train.reshape(len(y_train),1),y_pred_train1.reshape(len(y_pred_train1),1)),axis=1))
print()

print("For the test set:")
print("[Actual_observations Predicted_observations]")
print(np.concatenate((y_test.reshape(len(y_test),1),y_pred_test1.reshape(len(y_pred_test1),1)),axis=1))
```

<i>Accuracy of the model</i>

```python
from sklearn.metrics import accuracy_score, confusion_matrix

con_mat_lr_train=confusion_matrix(y_train,y_pred_train1)
con_mat_lr_test=confusion_matrix(y_test,y_pred_test1)

print("For the training set:")
print(f"[[True_Negative False_Positive]\t= {con_mat_lr_train[0]}")
print(f"[False_Negative True_Positive]]\t  {con_mat_lr_train[1]}")
print(f"Correct observations: {con_mat_lr_train[0][0]+con_mat_lr_train[1][1]}")
print(f"Incorrect observations: {con_mat_lr_train[0][1]+con_mat_lr_train[1][0]}")
print(f"Accuracy for the training set: {accuracy_score(y_train,y_pred_train1)*100}%")
print()

print("For the test set:")
print(f"[[True_Negative False_Positive]\t= {con_mat_lr_test[0]}")
print(f"[False_Negative True_Positive]]\t  {con_mat_lr_test[1]}")
print(f"Correct observations: {con_mat_lr_test[0][0]+con_mat_lr_test[1][1]}")
print(f"Incorrect observations: {con_mat_lr_test[0][1]+con_mat_lr_test[1][0]}")
print(f"Accuracy for the test set: {accuracy_score(y_test,y_pred_test1)*100}%")
```

```python
from sklearn.metrics import classification_report

print("Classification report for the training set:")
print(classification_report(y_train,y_pred_train1))
print()

print("Classification report for the test set:")
print(classification_report(y_test,y_pred_test1))
```

# Model Comparison — Phishing URL Detection

| Model | Train Accuracy | Test Accuracy | Cross-Val Accuracy (10-fold) | Precision (Test) | Recall (Test) | F1-Score (Test) |
|---|---|---|---|---|---|---|
| Logistic Regression | 85.66% | 85.60% | 85.66% | 0.86 | 0.86 | 0.85 |
| K-Nearest Neighbors (k=15) | 96.97% | 96.40% | — | 0.97 | 0.97 | 0.97 |
| Decision Tree (entropy) | 99.99% | 98.58% | 98.55% | 1.00 | 1.00 | 1.00 |
| Random Forest (n=500) | 99.99% | 99.09% | — | 1.00 | 1.00 | 1.00 |
| Gaussian Naive Bayes | 76.89% | 76.86% | 76.88% | 0.83 | 0.77 | 0.75 |

## Confusion Matrix Summary (Test Set)

| Model | True Negative | False Positive | False Negative | True Positive |
|---|---|---|---|---|
| Logistic Regression | 14,784 | 5,416 | 1,377 | 25,582 |
| K-Nearest Neighbors | 18,882 | 1,318 | 378 | 26,581 |
| Decision Tree | 19,841 | 359 | 310 | 26,649 |
| Random Forest | 19,946 | 254 | 177 | 26,782 |
| Gaussian Naive Bayes | 9,426 | 10,774 | 139 | 26,820 |

## Key Observations

- **Random Forest** performed best overall on the test set (99.09% accuracy), with the lowest combined false positive + false negative count.
- **Decision Tree** came close (98.58%), but with a slightly higher training-vs-test gap, suggesting mild overfitting compared to Random Forest.
- **KNN** gave solid results (96.40%) but is computationally expensive at prediction time compared to tree-based models.
- **Logistic Regression** and **Gaussian Naive Bayes** underperformed relative to tree-based models — likely because the decision boundary is non-linear and features have complex interactions that linear/probabilistic models can't capture as well.
- **Gaussian Naive Bayes** had the highest false positive count (10,774), meaning it incorrectly flagged many legitimate URLs as phishing — its independence assumption doesn't hold well for this feature set.

## Recommended Model
**Random Forest** is the best choice for deployment — highest test accuracy, best precision/recall balance, and more robust to overfitting than the single Decision Tree.
